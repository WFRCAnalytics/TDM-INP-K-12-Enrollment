```python
import pandas as pd
#import matplotlib.ticker as mtick
import numpy as np
```

```python
df1=pd.read_csv ('results/K12_Enrollment_public_MAZ.csv')
df2=pd.read_csv ('results/privK12_Enrollment_MAZ.csv')

# Combine them row-wise (default axis=0)
df_combined = pd.concat([df1, df2], ignore_index=True)
display (df_combined)
```

```python
# Group by CO_TAZID and sum the enrollment columns
df_grouped = df_combined.groupby('maz_id', as_index=False)[['Enrol_Elem', 'Enrol_Midl', 'Enrol_High']].sum()

display(df_grouped)
```

```python
#export CO_TAZID enrollment to csv

df_grouped.to_csv("results\GroupedEnrollmentByTAZ06252026.csv", index=False)
# Calculate totals one by one
total_elem = df_grouped['Enrol_Elem'].sum()
total_midl = df_grouped['Enrol_Midl'].sum()
total_high = df_grouped['Enrol_High'].sum()

print(f"Total Elementary Enrollment: {total_elem}")
print(f"Total Middle School Enrollment: {total_midl}")
print(f"Total High School Enrollment: {total_high}")
```
